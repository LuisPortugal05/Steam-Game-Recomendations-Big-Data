```python
import os
import json
from datetime import datetime

RAW_DIR = "data/raw/"
LOG_FILE = "data/ingestion_log.json"

def create_ingestion_log(directory, log_path):
    print(f"Scanning directory: {directory}...")
    log_data = []
    
    for filename in os.listdir(directory):
        if filename.endswith(".csv") or filename.endswith(".json"):
            filepath = os.path.join(directory, filename)
            file_stats = os.stat(filepath)
            
            file_info = {
                "file_name": filename,
                "ingestion_timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                "file_size_mb": round(file_stats.st_size / (1024 * 1024), 2),
                "status": "RAW_STORED_SUCCESSFULLY"
            }
            log_data.append(file_info)
            print(f"Logged: {filename} ({file_info['file_size_mb']} MB)")
            
    with open(log_path, "w") as log_file:
        json.dump(log_data, log_file, indent=4)
        
    print(f"\nIngestion log successfully created at: {log_path}")

create_ingestion_log(RAW_DIR, LOG_FILE)
```
