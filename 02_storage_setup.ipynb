```python
import psycopg2

DB_PARAMS = {
    "host": "postgres-db",
    "database": "steam_analytics",
    "user": "admin",
    "password": "admin"
}

def setup_curated_database():
    print("Connecting to PostgreSQL Database...")
    try:
        conn = psycopg2.connect(**DB_PARAMS)
        conn.autocommit = True
        cursor = conn.cursor()
        print("[SUCCESS] Connected to steam_analytics database!")

        create_table_query = """
        CREATE TABLE IF NOT EXISTS game_insights (
            app_id INTEGER PRIMARY KEY,
            title VARCHAR(255),
            price_final FLOAT,
            total_reviews INTEGER,
            positive_recommendations INTEGER,
            avg_hours_played FLOAT,
            recommendation_rate FLOAT
        );
        """
        
        cursor.execute(create_table_query)
        print("[SUCCESS] Table 'game_insights' is ready in the Curated Layer!")

        cursor.close()
        conn.close()

    except Exception as e:
        print(f"[ERROR] Could not connect to database. Error: {e}")

setup_curated_database()
```
