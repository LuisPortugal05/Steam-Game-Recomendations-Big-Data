```python
import os
import time
import pandas as pd

RAW_FILE = "data/raw/recommendations.csv"
STREAM_DIR = "data/streaming_input/"

def simulate_live_stream():
    print("Starting Live Steam Review Feed Simulation...")
    
    df = pd.read_csv(RAW_FILE, nrows=5000)
    
    os.makedirs(STREAM_DIR, exist_ok=True)
    
    chunk_size = 500
    for i in range(0, len(df), chunk_size):
        chunk = df.iloc[i:i+chunk_size]
        
        filename = f"{STREAM_DIR}/live_reviews_{int(time.time())}.csv"
        
        chunk.to_csv(filename, index=False)
        print(f"[LIVE] 🟢 Sent {len(chunk)} new reviews to {filename}")
        
        time.sleep(3)
        
    print("Simulation complete.")

simulate_live_stream()
```
